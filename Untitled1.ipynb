```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Настройка стиля
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

# =====================================================
# 1. ЗАГРУЗКА ДАННЫХ (Boston Housing)
# =====================================================
# Скачайте файл HousingData.csv с Kaggle и укажите путь
df = pd.read_csv('HousingData.csv')  # замените на ваш путь

print("=" * 60)
print("НАБОР ДАННЫХ: BOSTON HOUSING")
print("=" * 60)
print(f"Размерность: {df.shape[0]} строк, {df.shape[1]} столбцов")
print("\nПервые 5 строк:")
print(df.head())
print("\nОписание признаков:")
print("""
- CRIM: уровень преступности на душу населения
- ZN: доля жилой зоны > 25.000 кв.футов
- INDUS: доля не розничных торговых акров в городе
- CHAS: переменная Чарльза-Ривер (1 - рядом с рекой, 0 - нет)
- NOX: концентрация оксидов азота
- RM: среднее количество комнат в жилье
- AGE: доля занятых под жильё построек до 1940 года
- DIS: взвешенное расстояние до 5 центров занятости в Бостоне
- RAD: индекс доступности к радиальным шоссе
- TAX: полная ставка налога на имущество
- PTRATIO: соотношение учеников и учителей в городе
- B: 1000*(Bk - 0.63)^2, где Bk - доля чернокожих
- LSTAT: % населения с низким статусом
- MEDV: медианная стоимость домов (целевая переменная)
""")
```

```python
print("\n" + "=" * 60)
print("АНАЛИЗ ПРОПУСКОВ")
print("=" * 60)

missing_info = pd.DataFrame({
    'Кол-во пропусков': df.isnull().sum(),
    'Доля пропусков (%)': (df.isnull().sum() / len(df) * 100).round(2)
})
missing_info = missing_info[missing_info['Кол-во пропусков'] > 0].sort_values('Кол-во пропусков', ascending=False)
print(missing_info)

# Визуализация пропусков
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), yticklabels=False, cbar=True, cmap='viridis')
plt.title('Карта пропусков в датасете Boston Housing')
plt.show()
```

```python
print("\n" + "=" * 60)
print("ОБРАБОТКА КАТЕГОРИАЛЬНОГО ПРИЗНАКА")
print("=" * 60)

# Выбираем категориальный/бинарный признак с пропусками
# CHAS - бинарный (1 - рядом с рекой, 0 - нет), есть пропуски
categorical_col = 'CHAS'  # <-- ИСПРАВЛЕНО: заглавными буквами
print(f"Выбран признак: {categorical_col} (бинарный/категориальный)")
print(f"Уникальные значения до обработки: {df[categorical_col].unique()}")
print(f"Кол-во пропусков: {df[categorical_col].isnull().sum()}")

# Способ: заполнение МОДОЙ (наиболее частым значением)
mode_value = df[categorical_col].mode()[0]
df[categorical_col].fillna(mode_value, inplace=True)

print(f"\nСпособ обработки: заполнение МОДОЙ (значение {mode_value})")
print(f"Кол-во пропусков после обработки: {df[categorical_col].isnull().sum()}")
print(f"Распределение после обработки:\n{df[categorical_col].value_counts()}")
```

```python
print("\n" + "=" * 60)
print("ОБРАБОТКА КОЛИЧЕСТВЕННОГО ПРИЗНАКА")
print("=" * 60)

# Выбираем количественный признак с пропусками
numeric_col = 'CRIM'  # <-- ЗАГЛАВНЫМИ БУКВАМИ
print(f"Выбран признак: {numeric_col} (уровень преступности)")
print(f"Статистика ДО обработки:")
print(f"  Среднее: {df[numeric_col].mean():.4f}")
print(f"  Медиана: {df[numeric_col].median():.4f}")
print(f"  Стандартное отклонение: {df[numeric_col].std():.4f}")
print(f"Кол-во пропусков: {df[numeric_col].isnull().sum()}")

# Способ: заполнение МЕДИАНОЙ
median_value = df[numeric_col].median()
df[numeric_col].fillna(median_value, inplace=True)

print(f"\nСпособ обработки: заполнение МЕДИАНОЙ (значение {median_value:.4f})")
print(f"Кол-во пропусков после обработки: {df[numeric_col].isnull().sum()}")
print(f"Статистика ПОСЛЕ обработки:")
print(f"  Среднее: {df[numeric_col].mean():.4f}")
print(f"  Медиана: {df[numeric_col].median():.4f}")
```

```python
print("\n" + "=" * 60)
print("ПАРНЫЕ ДИАГРАММЫ (PAIRPLOT)")
print("=" * 60)

# Выбираем ключевые числовые признаки (ВСЕ ЗАГЛАВНЫМИ)
selected_features = ['CRIM', 'RM', 'AGE', 'DIS', 'TAX', 'LSTAT', 'MEDV']
df_selected = df[selected_features].copy()

# Строим pairplot
sns.pairplot(df_selected, diag_kind='kde', height=2.2)
plt.suptitle('Парные диаграммы основных признаков Boston Housing', y=1.02, fontsize=14)
plt.show()

# Альтернатива: pairplot с цветовой дифференциацией по признаку CHAS
sns.pairplot(df[selected_features + ['CHAS']], hue='CHAS', diag_kind='hist', palette='Set1', height=2)
plt.suptitle('Парные диаграммы с разделением по признаку CHAS', y=1.02, fontsize=14)
plt.show()
```

```python
print("ДОПОЛНИТЕЛЬНЫЕ ГРАФИКИ")
print("=" * 60)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Гистограмма целевой переменной MEDV
axes[0, 0].hist(df['MEDV'], bins=30, edgecolor='black', color='skyblue')
axes[0, 0].set_title('Распределение стоимости домов (MEDV)')
axes[0, 0].set_xlabel('MEDV')
axes[0, 0].set_ylabel('Частота')

# Ящик с усами для RM
axes[0, 1].boxplot(df['RM'].dropna())
axes[0, 1].set_title('Распределение количества комнат (RM)')
axes[0, 1].set_ylabel('RM')

# Scatter plot: RM vs MEDV
axes[1, 0].scatter(df['RM'], df['MEDV'], alpha=0.6, edgecolors='k')
axes[1, 0].set_title('Зависимость стоимости от количества комнат')
axes[1, 0].set_xlabel('RM')
axes[1, 0].set_ylabel('MEDV')

# Scatter plot: LSTAT vs MEDV
axes[1, 1].scatter(df['LSTAT'], df['MEDV'], alpha=0.6, color='green', edgecolors='k')
axes[1, 1].set_title('Зависимость стоимости от доли населения с низким статусом')
axes[1, 1].set_xlabel('LSTAT')
axes[1, 1].set_ylabel('MEDV')

plt.tight_layout()
plt.show()
```
